# Spacy

Spacy по умолчанию установлен в Colab!

```python
import spacy
```

Сначала нам нужна модель. Вот [список](https://spacy.io/models). Вспомним заслуги Александра Кукушкина и загрузим одну из его моделей. Напоминаю, что модели medium и large для русского языка включают в себя модели Navec. Модель small не позволяет выполнять векторизацию с использованием Spacy.

```python
! spacy download ru_core_news_lg -q
```

```python
model = spacy.load('ru_core_news_lg')
```

Создадим объект Doc

```python
text = 'Сегодня мы с вами будем изучать фреймворк Spacy! Я фанат Александра Кукушнина, а вы?'
doc = model(text)
doc, type(doc)
```

Doc - это абстракция над списком токенов.

```python
len(doc), len(text)
```

```python
doc[:10], doc[3], type(doc[3])
```

Для каждого токена мы можем получить много полезной информации

```python
doc[11].is_digit, doc[11].lower_, doc[11].lemma_, doc[11].has_vector, doc[11].ent_type_
```

А еще у нас есть объект Span

```python
doc[11:13].lemma_, type(doc[11:13])
```

Можем детектировать концы предложений

```python
list(doc.sents), type(next(doc.sents))
```

Стоп-слова

```python
model.Defaults.stop_words
```

Извлечение именованных сущностей

```python
for token in doc:
    print(f'{token.text} {token.ent_type_} {token.ent_iob_}')
```

Другой способ

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
from spacy import displacy
```

```python
displacy.render(doc, style='ent', jupyter=True)
```

Другая информация о токенах. Подробности здесь.

```python
for token in doc:
    print(f'{token.text}\t{token.morph}\t{token.dep_} {token.pos_}')
```

Поясни за терм!

```python
spacy.explain('nsubj'), spacy.explain('AUX')
```

```python
import spacy_sentence_bert
model2 = spacy_sentence_bert.load_model('xx_distiluse_base_multilingual_cased_v2')
```
