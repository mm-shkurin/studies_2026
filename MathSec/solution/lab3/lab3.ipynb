## 1. Подготовка текста

Текст приводится к тому же виду, что и для аффинного шифра: все символы в нижнем регистре, удаляются пробелы и знаки препинания. Учитываются только буквы русского алфавита, е/ё эквивалентны.

```python
import re
from collections import Counter
import math
import matplotlib.pyplot as plt

with open('text.txt', 'r', encoding='utf-8') as f:
    text = f.read()

text = re.sub(r'\[.*?\]', '', text)

text = text.lower()

text = text.replace('ё', 'е')

text = re.sub(r'[^а-я]', '', text)

print(f"Подготовленный текст ({len(text)} символов):")
print(text[:100] + "...")
```

## 2. Функция расчёта энтропии k-грамм

Энтропия k-грамм рассчитывается по формуле:
$$H_k(T) = -\sum_{i=1}^{n} p(z_i^k) \log_2(p(z_i^k)),$$

где $n$ — количество различных k-грамм, $p(z_i^k)$ — частота встречаемости k-граммы.

```python
def kgram_entropy(text, k):
    """
    Рассчитывает энтропию k-грамм для текста.
    
    Параметры:
        text: str — текст (без пробелов и знаков препинания)
        k: int — размер k-граммы
    
    Возвращает:
        float: энтропия H_k(T)
    """
    if k > len(text):
        return 0.0
    
    # Извлечение всех k-грамм
    kgrams = [text[i:i+k] for i in range(len(text) - k + 1)]
    
    if not kgrams:
        return 0.0
    
    # Подсчёт частот
    total = len(kgrams)
    freq = Counter(kgrams)
    
    # Расчёт энтропии
    entropy = 0.0
    for count in freq.values():
        p = count / total
        entropy -= p * math.log2(p)
    
    return entropy
```

## 3. Расчёт энтропии для k = 1, 2, 3, 4, 5

```python
# Расчёт энтропии для k = 1..5
results = []
for k in range(1, 6):
    Hk = kgram_entropy(text, k)
    Hk_div_k = Hk / k
    results.append((k, Hk, Hk_div_k))
    print(f"k = {k}: H_k = {Hk:.4f}, H_k/k = {Hk_div_k:.4f}")
```

## 4. Таблица результатов

```python
import pandas as pd

df = pd.DataFrame(results, columns=['k', 'H_k', 'H_k/k'])
df
```

## 5. График зависимости H_k/k от k

```python
plt.figure(figsize=(8, 5))
plt.plot(df['k'], df['H_k/k'], 'o-', linewidth=2, markersize=8)
plt.xlabel('k', fontsize=12)
plt.ylabel('H_k/k', fontsize=12)
plt.title('Зависимость H_k/k от k (вариант 22)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.xticks(range(1, 6))
plt.show()
```
